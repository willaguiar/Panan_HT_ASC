```python
#geographical vars for climatological HT calculation
yh = cc.querying.getvar(exp,'yh',session,n=1).compute()
dxu = cc.querying.getvar(exp,'dxCv',session,n=1).compute()
dyt = cc.querying.getvar(exp,'dyCu',session,n=1).compute()
dxCv = cc.querying.getvar(exp,'dxCv',session,n=1).isel(xh=slice(0,14400),yq=slice(1,2039)).compute()
dxCu = cc.querying.getvar(exp,'dxCu',session,n=1).isel(xq=slice(1,14401),yh=slice(0,2038)).compute()
dyCu = cc.querying.getvar(exp,'dyCu',session,n=1).isel(xq=slice(1,14401),yh=slice(0,2038)).compute()
dyCv = cc.querying.getvar(exp,'dyCv',session,n=1).isel(xh=slice(0,14400),yq=slice(1,2039)).compute()
	
```

```python
#Interpolating into the grids we want
volcello_v=p0025_HTmean.volcello.sel(yh=lat_range).rename({'yh':'yq'})
volcello_v['yq']=np.array(dxCv.yq)
dzt_v=(volcello_v/(dxCv*dyCv)).isel(z_l=slice(0,51)).compute()


volcello_u=p0025_HTmean.volcello.sel(yh=lat_range).rename({'xh':'xq'}); 
volcello_u['xq']=np.array(dxCu.xq)
dzt_u=(volcello_u/(dxCu*dyCu)).isel(z_l=slice(0,51))

#uhrho and vhrho calcupation step 1 
vhrho_nt=p0025_HTmean.vo*(dzt_v)
uhrho_et=p0025_HTmean.uo*(dzt_u)
```

```python

# horizontal widths are
dxu = dxu.sel(yq=lat_range)
dyt = dyt.sel(yh=lat_range)

#actual mass transport is
vhrho_nt = vhrho_nt*dxu
uhrho_et = uhrho_et*dyt
```

```python
# temperature set up 
yq = cc.querying.getvar(exp,'yq',session,n=1)
yq = yq.sel(yq=lat_range)
#Commented below as I'm not quite sure it needs to be done

tempx=p0025_HTmean.thetao.sel(yh=lat_range).interp(xh=uhrho_et.xq).interp(yh=uhrho_et.yh)
tempy=p0025_HTmean.thetao.sel(yh=lat_range).interp(xh=vhrho_nt.xh).interp(yh=vhrho_nt.yq)
	
```

```python
# climatological heat fluxes are
temp_xflux = (cp*rho_0*uhrho_et*tempx)
temp_yflux = (cp*rho_0*vhrho_nt*tempy)

#climatological freezing point fluxes are
freezing_point_heat_trans_zonal = (cp*rho_0*uhrho_et*temp_freezing)
freezing_point_heat_trans_meridional = (cp*rho_0*vhrho_nt*temp_freezing)
```
